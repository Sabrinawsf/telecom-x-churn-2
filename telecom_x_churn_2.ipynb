3.2.1 Modelo 1 — Regressão Logística

A Regressão Logística é um algoritmo amplamente utilizado em problemas de classificação binária, como a previsão de evasão de clientes (churn). Esse modelo estima a probabilidade de ocorrência de um evento com base nas variáveis explicativas.

**Necessidade de normalização**

A Regressão Logística é sensível à escala das variáveis, pois utiliza cálculos matemáticos que podem ser influenciados por diferenças de magnitude entre os atributos. Por esse motivo, foi aplicada padronização dos dados utilizando StandardScaler, garantindo que todas as variáveis contribuam de forma equilibrada para o modelo.

**Análise Crítica dos Modelos**

Após a avaliação das métricas, é possível comparar o desempenho dos modelos.

A **Regressão Logística** apresentou um desempenho consistente, sendo capaz de identificar padrões gerais nos dados. Como é um modelo mais simples e linear, tende a generalizar melhor, mas pode ter dificuldade em capturar relações mais complexas entre as variáveis.

O **Random Forest**, por outro lado, geralmente apresenta melhor desempenho em problemas de classificação, pois combina diversas árvores de decisão e consegue capturar relações não lineares entre as variáveis.

**Melhor desempenho**

O modelo que apresentar maior F1-score geralmente é considerado o mais adequado nesse tipo de problema, pois essa métrica equilibra precisão e recall, sendo especialmente útil em datasets desbalanceados como o de churn.

**Overfitting e Underfitting**

**Overfitting**

Pode ocorrer principalmente no Random Forest caso o modelo fique muito complexo. Isso acontece quando o modelo aprende muito bem os dados de treino, mas perde capacidade de generalizar para novos dados.

Possíveis ajustes:

* reduzir a profundidade das árvores
* diminuir o número de estimadores
* ajustar hiperparâmetros

**Underfitting**

Pode ocorrer na Regressão Logística caso o modelo seja simples demais para capturar os padrões do dataset.

Possíveis ajustes:

* adicionar novas variáveis
* ajustar hiperparâmetros
* testar modelos mais complexos

**Conclusão**

A comparação entre os dois modelos permite identificar qual abordagem apresenta melhor capacidade de prever a evasão de clientes. Em muitos casos, modelos baseados em árvores como o Random Forest tendem a apresentar melhor desempenho, enquanto a Regressão Logística oferece maior interpretabilidade e simplicidade.

Essa análise ajuda a escolher o modelo mais adequado para prever churn e apoiar estratégias de retenção de clientes.

## Principais fatores que influenciam a evasão de clientes e estratégias de retenção

A partir da análise exploratória dos dados e dos resultados obtidos pelos modelos preditivos, foi possível identificar alguns fatores que apresentam maior influência na evasão de clientes (churn). Esses fatores ajudam a compreender o comportamento dos clientes e permitem propor estratégias mais eficazes para reduzir cancelamentos.

### Principais fatores associados à evasão

**1. Tempo de contrato (tenure)**
Clientes com menor tempo de contrato tendem a apresentar maior probabilidade de cancelar o serviço. Isso indica que muitos cancelamentos ocorrem nos primeiros meses de utilização, possivelmente devido a expectativas não atendidas ou dificuldades iniciais na adaptação ao serviço.

**2. Valor mensal do serviço (charges_monthly)**
Clientes com mensalidades mais altas podem demonstrar maior propensão à evasão, especialmente quando percebem que o custo não corresponde ao valor ou à qualidade do serviço oferecido.

**3. Total gasto pelo cliente (charges_total)**
O valor total gasto está fortemente relacionado ao tempo de permanência do cliente. Clientes com menor valor acumulado geralmente são aqueles que permanecem menos tempo na empresa e apresentam maior probabilidade de cancelamento.

**4. Tipo de contrato**
Clientes com contratos mensais tendem a cancelar com mais frequência do que aqueles que possuem contratos de longo prazo. A ausência de compromisso contratual facilita a decisão de cancelar o serviço.

**5. Método de pagamento**
Alguns métodos de pagamento podem estar associados a maior taxa de evasão, indicando possíveis dificuldades no processo de pagamento ou menor vínculo com a empresa.

---

### Estratégias de retenção baseadas nos resultados

Com base nos fatores identificados, algumas estratégias podem ser implementadas para reduzir a evasão de clientes.

**1. Fortalecer a experiência inicial do cliente**
Como muitos cancelamentos ocorrem nos primeiros meses, é importante melhorar o processo de onboarding. Isso pode incluir suporte inicial mais ativo, acompanhamento nos primeiros meses e orientações claras sobre o uso dos serviços.

**2. Incentivar contratos de longo prazo**
Oferecer benefícios para clientes que optam por contratos anuais ou de maior duração pode ajudar a reduzir a taxa de cancelamento. Descontos progressivos, vantagens exclusivas ou bônus de fidelidade podem estimular a permanência.

**3. Ajustar estratégias de preço e valor percebido**
Para clientes com mensalidades mais altas, é importante reforçar a percepção de valor do serviço. Isso pode ser feito por meio de melhorias na qualidade do atendimento, comunicação dos benefícios do plano ou ofertas personalizadas.

**4. Identificar clientes com risco de churn**
Utilizando os modelos preditivos desenvolvidos, a empresa pode identificar clientes com maior probabilidade de cancelamento e direcionar ações específicas de retenção, como ofertas especiais ou atendimento prioritário.

**5. Melhorar serviços complementares e suporte técnico**
A oferta de serviços adicionais e suporte eficiente pode aumentar o nível de satisfação do cliente e reduzir a probabilidade de evasão.

---

### Conclusão

A análise dos dados e dos modelos preditivos permitiu identificar padrões importantes associados ao churn. Variáveis relacionadas ao tempo de contrato, custos e características do serviço demonstraram forte influência na decisão de cancelamento.

Com base nesses resultados, a empresa pode implementar estratégias direcionadas para melhorar a experiência do cliente, aumentar o engajamento e reduzir a evasão, contribuindo para maior retenção e sustentabilidade do negócio.
